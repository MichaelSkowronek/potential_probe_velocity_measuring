### Imports

```python
from pathlib import Path


from ppvm_py.utils import save_object
from ppvm_py.utils import load_object

from ppvm_py.data_processing.patt_fld8v import parse_tecplot_3d_timeseries
from ppvm_py.data_processing.patt_fld8v import print_xyz_slice_example_from_parsed_data
```

### General Args

```python
# IMPORTANT: Restart Kernel when swapping datasets to reload them!
Ha = 1000

if Ha == 300:
    patt_fld8v_folder_path = Path("../../../../data/4pi_re1000_ha300.360_ch28.p1/patt_fld8v/")
elif Ha == 1000:
    patt_fld8v_folder_path = Path("../../../../data/4pi_re1000_ha1000.384.pi/patt_fld8v/")

path = patt_fld8v_folder_path / Path("patt_fld8v.dat")

print_loaded_data_example_ = True
delete_files = False
```

```python
from ppvm_py.data_processing.utils import truncate_file_at_line

# path = truncate_file_at_line(
#     filepath=path,
#     line_number=5438404,
#     )

path = "C:/Users/hydro/Documents/phd/nuclear_fusion_cooling/data/4pi_re1000_ha1000.384.pi/patt_fld8v/patt_fld8v_truncated.dat"
```

### Load

```python
data = {}
(
    data['timeseries'],
    data['coord_to_index_map'],
    data['labels'],
    ) = parse_tecplot_3d_timeseries(path)
if print_loaded_data_example_ is True:
    print_xyz_slice_example_from_parsed_data(data)
```

### Store

```python
pickle_dir = Path("pkl")
pickle_path = patt_fld8v_folder_path / pickle_dir / Path("patt_fld8v.pkl")
```

```python
save_object(data, pickle_path)
```

### Verification

```python
loaded_data = load_object(pickle_path)
print_xyz_slice_example_from_parsed_data(loaded_data)
```

### Delete

```python
if delete_files is True:
    abs_path = path.resolve()
    path.unlink()
    print(f"Successfully removed {abs_path}")
```
