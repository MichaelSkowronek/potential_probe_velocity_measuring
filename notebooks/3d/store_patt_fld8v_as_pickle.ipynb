### General Args

```python
# IMPORTANT: Restart Kernel when swapping datasets to reload them!
Ha = 300

if Ha == 300:
    patt_fld8v_folder_path = Path("../../../../data/4pi_re1000_ha300.360_ch28.p1/patt_fld8v/")
elif Ha == 1000:
    patt_fld8v_folder_path = Path("../../../../data/4pi_re1000_ha1000.384.pi/patt_fld8v/")
data_path = patt_fld8v_folder_path / Path("patt_fld8v.dat")

pkl_store_file_path = patt_fld8v_folder_path / Path("pkl/patt_fld8v.pkl")

print_loaded_data_example_ = True
delete_files = False
```

### Load

```python
data = {}
(
    data['timeseries'],
    data['coord_to_index_map'],
    data['labels'],
    ) = parse_tecplot_3d_timeseries(data_path)
if print_loaded_data_example_ is True:
    print_xyz_slice_example_from_parsed_data(data)
```

### Store

```python
save_object(data, pkl_store_file_path)
```

### Verification

```python
loaded_data = load_object(pkl_store_file_path)
print_xyz_slice_example_from_parsed_data(loaded_data)
```

### Delete

```python
if delete_files is True:
    abs_path = data_path.resolve()
    data_path.unlink()
    print(f"Successfully removed {abs_path}")
```
